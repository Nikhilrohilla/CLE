Read The CSV


```python
# Step 1: Read the Yellow Taxi CSV

from pyspark.sql import functions as F

file_path = "/Volumes/workspace/default/yellow_taxi_volume/yellow_tripdata_2016-01.csv"

df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .option("mode", "PERMISSIVE")
    .csv(file_path)
)

print("CSV successfully loaded!")
```

```python
# Display first 5 records
display(df.limit(5))

# Display column names
print("Total Columns:", len(df.columns))
print("Column Names:", df.columns)

# Display schema
df.printSchema()
```

DataSet Overview


```python
# Step 2: Dataset overview

# Total number of records
total_rows = df.count()

print("========== DATASET OVERVIEW ==========")
print("Total Rows:", total_rows)
print("Total Columns:", len(df.columns))

# Inspect important columns
display(
    df.select(
        "VendorID",
        "tpep_pickup_datetime",
        "tpep_dropoff_datetime",
        "passenger_count",
        "trip_distance",
        "fare_amount",
        "tip_amount",
        "total_amount"
    ).limit(10)
)
```

Display Null Counts

```python
from pyspark.sql import functions as F

# Count NULL values in every column
null_counts = df.select([
    F.count(
        F.when(F.col(c).isNull(), c)
    ).alias(c)
    for c in df.columns
])

display(null_counts)
```

Count Duplicate

```python
# Count duplicate rows
total_records = df.count()
unique_records = df.dropDuplicates().count()

duplicate_count = total_records - unique_records

print("Total Records:", total_records)
print("Unique Records:", unique_records)
print("Duplicate Records:", duplicate_count)
```

Remove Duplicate

```python
# Remove exact duplicate records

df_deduplicated = df.dropDuplicates()

print("Original records:", df.count())
print("Records after deduplication:", df_deduplicated.count())
print("Duplicates removed:", df.count() - df_deduplicated.count())
```

Clean column names lowercase, no spaces

```python
import re

# Clean column names: lowercase and snake_case

def clean_column_name(column_name):
    # Insert underscore before uppercase letters
    name = re.sub(r'([a-z0-9])([A-Z])', r'\1_\2', column_name)

    # Convert to lowercase
    name = name.lower()

    # Replace spaces and special characters with underscores
    name = re.sub(r'[^a-z0-9]+', '_', name)

    # Remove leading/trailing underscores
    return name.strip('_')


# Rename all columns
df_clean_names = df_deduplicated.toDF(
    *[clean_column_name(c) for c in df_deduplicated.columns]
)

print("Cleaned Column Names:")
print(df_clean_names.columns)

df_clean_names.printSchema()
```

Fix types numbers to float


```python
from pyspark.sql.types import (
    IntegerType,
    DoubleType,
    TimestampType
)

# Define expected data types
type_mapping = {
    "vendor_id": IntegerType(),
    "passenger_count": IntegerType(),
    "ratecode_id": IntegerType(),
    "payment_type": IntegerType(),
    "tpep_pickup_datetime": TimestampType(),
    "tpep_dropoff_datetime": TimestampType(),
    "trip_distance": DoubleType(),
    "pickup_longitude": DoubleType(),
    "pickup_latitude": DoubleType(),
    "dropoff_longitude": DoubleType(),
    "dropoff_latitude": DoubleType(),
    "fare_amount": DoubleType(),
    "extra": DoubleType(),
    "mta_tax": DoubleType(),
    "tip_amount": DoubleType(),
    "tolls_amount": DoubleType(),
    "improvement_surcharge": DoubleType(),
    "total_amount": DoubleType()
}

# Apply type conversions
df_typed = df_clean_names

for column, data_type in type_mapping.items():
    df_typed = df_typed.withColumn(
        column,
        F.col(column).cast(data_type)
    )

print("Data types updated successfully!")
df_typed.printSchema()
```

Check no negative amounts, else stop with an error

```python
# Step 6: Validate negative financial amounts

financial_columns = [
    "fare_amount",
    "extra",
    "mta_tax",
    "tip_amount",
    "tolls_amount",
    "improvement_surcharge",
    "total_amount"
]

# Count negative values in each financial column
negative_counts = df_typed.select([
    F.sum(
        F.when(F.col(c) < 0, 1).otherwise(0)
    ).alias(c)
    for c in financial_columns
])

print("Negative Amount Validation:")
display(negative_counts)
```

```python
# Collect negative value counts
negative_result = negative_counts.collect()[0].asDict()

print("Negative Amount Validation Results:")

for column, count in negative_result.items():
    print(f"{column}: {count}")

# Stop the pipeline if negative values exist
total_negative = sum(negative_result.values())

if total_negative > 0:
    raise ValueError(
        f"DATA VALIDATION FAILED: Found {total_negative} negative financial values."
    )

print("Validation successful! No negative amounts found.")
```

```python
import logging

# Configure logging
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logger = logging.getLogger("YellowTaxiPipeline")

try:
    logger.info("Starting financial validation")

    # Collect negative-value counts
    negative_result = negative_counts.collect()[0].asDict()

    total_negative = sum(negative_result.values())

    logger.info("Financial validation completed")
    logger.info(f"Total negative-value occurrences: {total_negative}")

    for column, count in negative_result.items():
        logger.info(f"{column}: {count} negative values")

    # Enforce validation rule
    if total_negative > 0:
        raise ValueError(
            f"Validation failed: {total_negative} negative-value occurrences found"
        )

    logger.info("Validation successful. Dataset is ready.")

except Exception:
    logger.exception("YELLOW TAXI PIPELINE FAILED")
    raise
```

Records containing negative financial values

```python
from pyspark.sql import functions as F

financial_columns = [
    "fare_amount",
    "extra",
    "mta_tax",
    "tip_amount",
    "tolls_amount",
    "improvement_surcharge",
    "total_amount"
]

negative_condition = None

for column in financial_columns:
    condition = F.col(column) < 0
    negative_condition = (
        condition if negative_condition is None
        else negative_condition | condition
    )

negative_records = df_typed.filter(negative_condition)

print("Records containing at least one negative financial value:")
print(negative_records.count())

display(negative_records.limit(10))
```

Store rejected records separately

```python
# Save records that failed financial validation
quarantine_path = (
    "/Volumes/workspace/default/yellow_taxi_volume/"
    "quarantine/negative_records"
)

negative_records.write \
    .mode("overwrite") \
    .parquet(quarantine_path)

print("Quarantine records saved successfully!")
print("Rejected records:", negative_records.count())
print("Location:", quarantine_path)
```

Separate Valid Taxi Records From rejected records and verify that no records are lost during the cleaning process.

```python
# Keep records with no negative financial values
clean_df = df_typed.filter(~negative_condition)

# Count valid records
clean_count = clean_df.count()
original_count = df_typed.count()
rejected_count = negative_records.count()

print("DATA QUALITY SUMMARY")
print("--------------------")
print("Original records:", original_count)
print("Rejected records:", rejected_count)
print("Valid records:", clean_count)
print("Reconciliation:", clean_count + rejected_count == original_count)
```

Validate the candidate clean dataset

```python
from pyspark.sql import functions as F

# Check negative financial values
remaining_negative = clean_df.filter(
    negative_condition
).count()

# Check null values in financial columns
null_checks = clean_df.select([
    F.sum(
        F.when(F.col(c).isNull(), 1).otherwise(0)
    ).alias(c)
    for c in financial_columns
]).collect()[0].asDict()

print("FINAL DATA QUALITY CHECK")
print("------------------------")
print("Remaining negative records:", remaining_negative)
print("Null counts:", null_checks)

# Validation gate
if remaining_negative > 0:
    raise ValueError("Negative values still exist!")

if any(count > 0 for count in null_checks.values()):
    raise ValueError("Null financial values still exist!")

print("All financial validation checks passed!")
```

Save the cleaned dataset as Parquet

```python
# Define output location directly inside the volume
output_path = "/Volumes/workspace/default/yellow_taxi_volume/yellow_taxi_clean"

# Save cleaned DataFrame as Parquet
clean_df.write \
    .mode("overwrite") \
    .parquet(output_path)

print("SUCCESS: Parquet file saved!")
print("Records saved:", clean_df.count())
print("Output location:", output_path)
```
